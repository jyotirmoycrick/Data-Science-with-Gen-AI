8.  **What is a hash table, and how does it relate to dictionaries in Python?**

        In Python, a dict (dictionary) is an implementation of a hash table. When you do:
        ```python
        my_dict = {'apple': 3}
        ```
        Python hashes the key 'apple' to find where to store or retrieve its value efficiently.

10. **Explain why strings are immutable in Python?**

    * Strings are immutable in Python to:
        * Improve performance: Immutable objects can be cached and reused, saving memory.
        * Ensure safety: Since strings can't change, they can be safely used as dictionary keys or in sets.
        * Enable thread-safety: No risk of strings being changed by another thread during execution.
        Instead of changing a string, Python creates a new string when you modify it.
